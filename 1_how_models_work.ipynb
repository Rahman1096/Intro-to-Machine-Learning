## What is a Decision Tree?

Machine learning models learn patterns from data. One of the simplest and most intuitive models is a **Decision Tree**.

A decision tree asks a series of questions to reach a prediction. For example, to predict whether a house is expensive:

```
Does the house have more than 2 bedrooms?
    Yes → Is the house larger than 2000 sq ft?
              Yes → Expensive
              No  → Moderate
    No  → Affordable
```

The points at the bottom of the tree where predictions are made are called **leaves**.

## Fitting / Training a Model

**Fitting** (or **training**) a model means capturing patterns from data. The dataset used to fit the model is called **training data**.

After a model has been fit, you can apply it to new data to make **predictions**.

## Improving the Model

- A tree with more **splits** can capture more complex patterns.
- However, too many splits can cause **overfitting** (covered in Lesson 5).

## Summary

- Machine learning models learn patterns from data.
- A **Decision Tree** is one of the most intuitive models.
- **Fitting** a model means using training data to capture patterns.
- In the next lessons we will explore data and build our first model.

````python
# A simple illustration of decision tree logic

def predict_house_price(bedrooms, size_sqft):
    """Simple decision tree logic for house price category."""
    if bedrooms > 2:
        if size_sqft > 2000:
            return 'Expensive'
        else:
            return 'Moderate'
    else:
        return 'Affordable'

# Test the function
examples = [
    (4, 2500),
    (3, 1800),
    (2, 1200),
]

for bedrooms, size in examples:
    category = predict_house_price(bedrooms, size)
    print(f"Bedrooms: {bedrooms}, Size: {size} sqft → {category}")
````

## Key Takeaways

1. Decision trees split data based on feature values.
2. Each split creates a new **branch**; the end points are **leaves**.
3. The model **learns** the best splits from training data.
4. More splits can capture more patterns, but also risk **overfitting**.
