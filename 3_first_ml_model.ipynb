# Lesson 3: Your First Machine Learning Model

## Introduction

Now we will build our **first machine learning model**! We'll use scikit-learn (`sklearn`), the most popular library for building ML models.

## Steps to Building a Model

1. **Define** — What type of model? What parameters?
2. **Fit** — Capture patterns from training data.
3. **Predict** — Make predictions.
4. **Evaluate** — Determine how accurate the model's predictions are.

## Selecting the Prediction Target

The column we want to predict is called the **target** (often called `y`).

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Sample Melbourne housing data
data = {
    'Rooms': [2, 2, 3, 3, 4, 4, 2, 3, 4, 2],
    'Distance': [2.5, 2.5, 2.5, 2.5, 2.5, 3.0, 3.0, 3.5, 4.0, 4.5],
    'Bedroom2': [2, 2, 3, 3, 3, 4, 2, 3, 4, 2],
    'Bathroom': [1, 1, 2, 2, 1, 2, 1, 2, 2, 1],
    'Landsize': [202, 156, 134, 94, 120, 181, 245, 256, 300, 190],
    'BuildingArea': [79.0, 79.0, 150.0, 142.0, 210.0, 175.0, 95.0, 160.0, 220.0, 85.0],
    'YearBuilt': [1900, 1900, 1900, 2014, 2014, 2010, 1950, 1970, 2000, 1980],
    'Price': [1480000, 1035000, 1465000, 850000, 1600000, 1876000, 1636000, 1097000, 1350000, 750000]
}

melbourne_data = pd.DataFrame(data)

# Drop rows with missing values
melbourne_data = melbourne_data.dropna()

# Select the target
y = melbourne_data.Price
print("Target (Price):")
print(y)
```

## Choosing Features

The columns used as inputs to the model are called **features** (often called `X`).

```python
# Select features
feature_columns = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 'YearBuilt', 'Distance']

X = melbourne_data[feature_columns]
print("Features (X):")
print(X.describe())
```

## Building and Fitting the Model

```python
# Define the model
# random_state ensures reproducibility
melbourne_model = DecisionTreeRegressor(random_state=1)

# Fit the model (learn from the data)
melbourne_model.fit(X, y)

print("Model fitted successfully!")
```

## Making Predictions

```python
print("Making predictions for the first 5 houses:")
print(X.head())
print()
print("Predicted prices:")
print(melbourne_model.predict(X.head()))
print()
print("Actual prices:")
print(y.head().values)
```

## Summary

- The **target** (`y`) is the column we want to predict.
- **Features** (`X`) are the input columns used to make predictions.
- Use `DecisionTreeRegressor` from scikit-learn to create a model.
- Use `.fit(X, y)` to train the model.
- Use `.predict(X)` to make predictions.

> ⚠️ We made predictions on training data — this is not a valid evaluation. See the next lesson for proper model evaluation.
