## Exploring the Data

### `describe()` — Summary Statistics

The `describe()` method shows summary statistics for all numeric columns.

```python
melbourne_data.describe()
```

The output includes:
- **count** — Number of non-missing values
- **mean** — Average value
- **std** — Standard deviation (how spread out the values are)
- **min / max** — Minimum and maximum values
- **25% / 50% / 75%** — Quartiles

### `head()` — First Rows

```python
# See the first 5 rows
melbourne_data.head()
```

### Column Names

```python
# Show all column names
print(melbourne_data.columns.tolist())
```

### Selecting a Column

You can select a single column using dot notation or bracket notation.

```python
# Select the Price column
print(melbourne_data.Price)

# Equivalent:
print(melbourne_data['Price'])
```

## Summary

- Use `pd.read_csv()` to load data.

- Use `.head()` to preview rows.
- Use `.columns` to list all features.
- Select columns with `df['column_name']` or `df.column_name`.
