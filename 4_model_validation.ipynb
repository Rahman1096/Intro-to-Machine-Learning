# Lesson 4: Model Validation

## Introduction

You built a model — but how good is it?

We need a way to **measure model quality** to iteratively improve it.

## What is Model Validation?

A common mistake is evaluating a model using the same data it was trained on. This gives misleadingly good scores because the model has already "seen" that data.

**The solution**: split data into **training** and **validation** sets.

- **Training data** — used to fit the model.
- **Validation data** — held out, used only to evaluate the model.

## Mean Absolute Error (MAE)

We measure prediction quality using **Mean Absolute Error (MAE)**:

```
Error = Actual − Predicted
MAE  = mean(|Error|)
```

On average, our predictions are off by `MAE` dollars (for house prices).

````python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# Recreate the dataset from the previous lesson
data = {
    'Rooms': [2, 2, 3, 3, 4, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3],
    'Distance': [2.5, 2.5, 2.5, 2.5, 2.5, 3.0, 3.0, 3.5, 4.0, 4.5,
                 2.8, 3.2, 3.7, 4.1, 2.6, 2.9, 3.3, 3.8, 4.2, 2.7],
    'Bedroom2': [2, 2, 3, 3, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3],
    'Bathroom': [1, 1, 2, 2, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2],
    'Landsize': [202, 156, 134, 94, 120, 181, 245, 256, 300, 190,
                 210, 175, 230, 265, 310, 185, 240, 270, 195, 220],
    'BuildingArea': [79.0, 79.0, 150.0, 142.0, 210.0, 175.0, 95.0, 160.0, 220.0, 85.0,
                     130.0, 185.0, 100.0, 165.0, 225.0, 90.0, 140.0, 195.0, 80.0, 155.0],
    'YearBuilt': [1900, 1900, 1900, 2014, 2014, 2010, 1950, 1970, 2000, 1980,
                  1990, 2005, 1960, 1975, 2008, 1955, 1985, 2012, 1945, 1995],
    'Price': [1480000, 1035000, 1465000, 850000, 1600000, 1876000, 1636000, 1097000, 1350000, 750000,
              1200000, 1550000, 900000, 1100000, 1700000, 800000, 1300000, 1450000, 700000, 1050000]
}

melbourne_data = pd.DataFrame(data)
melbourne_data = melbourne_data.dropna()

feature_columns = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 'YearBuilt', 'Distance']
X = melbourne_data[feature_columns]
y = melbourne_data.Price
````

## Calculating MAE on Training Data (Naive Approach)

````python
# Fit model on ALL data and check error on same data
model = DecisionTreeRegressor(random_state=1)
model.fit(X, y)

predicted_prices = model.predict(X)
mae_in_sample = mean_absolute_error(y, predicted_prices)
print(f'In-sample MAE: ${mae_in_sample:,.0f}')
print('(This is misleadingly low - the model saw this data during training!)')
````

## Using a Validation Set (Correct Approach)

````python
# Split data: 80% training, 20% validation
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

print(f'Training samples: {len(train_X)}')
print(f'Validation samples: {len(val_X)}')

# Fit model on TRAINING data only
model = DecisionTreeRegressor(random_state=1)
model.fit(train_X, train_y)

# Evaluate on VALIDATION data
val_predictions = model.predict(val_X)
mae_out_of_sample = mean_absolute_error(val_y, val_predictions)
print(f'\nOut-of-sample MAE: ${mae_out_of_sample:,.0f}')
print('(This is a realistic measure of model performance on new data.)')
````

## Interpreting MAE

- An MAE of \$250,000 on houses that typically sell for \$1M–\$2M means predictions are off by about 15–25%.
- Lower MAE = better model.

## Summary

- Always evaluate your model on **data it has not seen** during training.
- Use `train_test_split` to create training and validation sets.
- **Mean Absolute Error (MAE)** is a simple, intuitive metric for regression.
