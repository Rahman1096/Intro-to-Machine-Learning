# Lesson 6: Random Forests

## Introduction

Decision trees leave you with a difficult decision. A deep tree with lots of leaves will overfit, but a shallow tree with few leaves will underfit.

Many modern machine learning algorithms perform better than a single decision tree. The **Random Forest** is one such algorithm, and it is a great starting point for many problems.

## What is a Random Forest?

A **Random Forest** builds many decision trees and averages their predictions. It uses two types of randomness to make diverse trees:

1. **Bootstrap sampling** — Each tree is trained on a random subset (with replacement) of the training data.
2. **Feature randomness** — At each split, only a random subset of features is considered.

By averaging many different trees, the random forest reduces the variance of the prediction and avoids overfitting.

```
Training Data
     │
     ├──► Tree 1 ──► Prediction 1
     ├──► Tree 2 ──► Prediction 2   ──► Average ──► Final Prediction
     ├──► Tree 3 ──► Prediction 3
     └──► Tree N ──► Prediction N
```

````python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# Recreate the dataset
data = {
    'Rooms': [2, 2, 3, 3, 4, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3,
              2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3],
    'Distance': [2.5, 2.5, 2.5, 2.5, 2.5, 3.0, 3.0, 3.5, 4.0, 4.5,
                 2.8, 3.2, 3.7, 4.1, 2.6, 2.9, 3.3, 3.8, 4.2, 2.7,
                 5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.5,
                 10.0, 10.5, 11.0, 11.5, 12.0, 12.5, 13.0, 13.5, 14.0, 14.5],
    'Bathroom': [1, 1, 2, 2, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2,
                 1, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2, 2, 1, 1, 2, 1, 2, 2, 1, 2],
    'Landsize': [202, 156, 134, 94, 120, 181, 245, 256, 300, 190,
                 210, 175, 230, 265, 310, 185, 240, 270, 195, 220,
                 180, 160, 200, 220, 250, 280, 320, 190, 210, 175,
                 300, 340, 260, 290, 310, 170, 230, 260, 285, 195],
    'BuildingArea': [79, 79, 150, 142, 210, 175, 95, 160, 220, 85,
                     130, 185, 100, 165, 225, 90, 140, 195, 80, 155,
                     110, 135, 175, 90, 155, 210, 85, 120, 170, 95,
                     200, 240, 145, 180, 215, 88, 132, 178, 200, 105],
    'YearBuilt': [1900, 1900, 1900, 2014, 2014, 2010, 1950, 1970, 2000, 1980,
                  1990, 2005, 1960, 1975, 2008, 1955, 1985, 2012, 1945, 1995,
                  2003, 1968, 1992, 2010, 1982, 1957, 2015, 1972, 1998, 2007,
                  1963, 2001, 1988, 1977, 2016, 1948, 1993, 2009, 1985, 1971],
    'Price': [1480000, 1035000, 1465000, 850000, 1600000, 1876000, 1636000, 1097000, 1350000, 750000,
              1200000, 1550000, 900000, 1100000, 1700000, 800000, 1300000, 1450000, 700000, 1050000,
              1150000, 980000, 1380000, 820000, 1250000, 1500000, 720000, 1010000, 1320000, 880000,
              1600000, 1800000, 1120000, 1280000, 1520000, 760000, 1180000, 1420000, 1080000, 930000]
}

melbourne_data = pd.DataFrame(data)
feature_columns = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 'YearBuilt', 'Distance']
X = melbourne_data[feature_columns]
y = melbourne_data.Price

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)
````

## Comparing Decision Tree vs Random Forest

````python
from sklearn.tree import DecisionTreeRegressor

# Decision Tree
dt_model = DecisionTreeRegressor(random_state=1)
dt_model.fit(train_X, train_y)
dt_preds = dt_model.predict(val_X)
dt_mae = mean_absolute_error(val_y, dt_preds)

# Random Forest
rf_model = RandomForestRegressor(n_estimators=100, random_state=1)
rf_model.fit(train_X, train_y)
rf_preds = rf_model.predict(val_X)
rf_mae = mean_absolute_error(val_y, rf_preds)

print(f"Decision Tree MAE:  ${dt_mae:>12,.0f}")
print(f"Random Forest MAE:  ${rf_mae:>12,.0f}")
improvement = (dt_mae - rf_mae) / dt_mae * 100
print(f"\nRandom Forest improves MAE by {improvement:.1f}%")
````

## Key Parameters

| Parameter | Description | Default |
|-----------|-------------|--------|
| `n_estimators` | Number of trees in the forest | 100 |
| `max_features` | Max features to consider at each split | `'sqrt'` |
| `max_depth` | Maximum depth of each tree | `None` (unlimited) |
| `random_state` | Seed for reproducibility | `None` |

````python
# Try different numbers of trees
for n_trees in [10, 50, 100, 200]:
    model = RandomForestRegressor(n_estimators=n_trees, random_state=1)
    model.fit(train_X, train_y)
    preds = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds)
    print(f"n_estimators={n_trees:>4}  |  MAE: ${mae:>12,.0f}")
````

## Summary

- A **Random Forest** combines many decision trees by averaging their predictions.
- It generally performs much better than a single decision tree.
- The `n_estimators` parameter controls how many trees to build — more trees usually means better performance, up to a point.
- Random forests work well with default parameters, making them a great baseline model.
