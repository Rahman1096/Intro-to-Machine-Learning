## The Competition Task

In the [Housing Prices Competition for Kaggle Learn Users](https://www.kaggle.com/c/home-data-for-ml-course), you are given:

- **train.csv** — Training data with features and the target (SalePrice).
- **test.csv** — Test data with features but **no target**.

Your goal is to predict the `SalePrice` for each house in the test set and submit a CSV file.

## Typical Workflow

1. Load training and test data.
2. Engineer features and handle missing values.
3. Train a model on the training data.
4. Make predictions on the test data.
5. Save predictions to a CSV file and submit.

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# ─────────────────────────────────────────────────
# In a real Kaggle notebook, load data like this:
#   train_data = pd.read_csv('../input/home-data-for-ml-course/train.csv', index_col='Id')
#   test_data  = pd.read_csv('../input/home-data-for-ml-course/test.csv',  index_col='Id')
# ─────────────────────────────────────────────────
# We simulate the data here for illustration:

import numpy as np
rng = np.random.default_rng(42)
n_train, n_test = 200, 50

def make_price(rooms, year, size):
    return (rooms * 150000 + (year - 1900) * 3000 + size * 300
            + rng.integers(-50000, 50000, size=len(rooms)))

rooms_tr = rng.integers(2, 6, size=n_train)
year_tr  = rng.integers(1900, 2020, size=n_train)
size_tr  = rng.integers(80, 300, size=n_train)

train_data = pd.DataFrame({
    'Id': range(1, n_train + 1),
    'OverallQual': rng.integers(1, 11, size=n_train),
    'GrLivArea': size_tr,
    'GarageCars': rng.integers(0, 4, size=n_train),
    'TotalBsmtSF': rng.integers(500, 2000, size=n_train),
    'FullBath': rng.integers(1, 4, size=n_train),
    'YearBuilt': year_tr,
    'Rooms': rooms_tr,
    'SalePrice': make_price(rooms_tr, year_tr, size_tr),
}).set_index('Id')

rooms_te = rng.integers(2, 6, size=n_test)
year_te  = rng.integers(1900, 2020, size=n_test)
size_te  = rng.integers(80, 300, size=n_test)

test_data = pd.DataFrame({
    'Id': range(n_train + 1, n_train + n_test + 1),
    'OverallQual': rng.integers(1, 11, size=n_test),
    'GrLivArea': size_te,
    'GarageCars': rng.integers(0, 4, size=n_test),
    'TotalBsmtSF': rng.integers(500, 2000, size=n_test),
    'FullBath': rng.integers(1, 4, size=n_test),
    'YearBuilt': year_te,
    'Rooms': rooms_te,
}).set_index('Id')

print(f"Training rows: {len(train_data)}, Test rows: {len(test_data)}")
print("\nTraining data preview:")
train_data.head()
```

## Step 1 — Prepare the Data

```python
# Select features (must exist in both train and test!)
features = ['OverallQual', 'GrLivArea', 'GarageCars',
            'TotalBsmtSF', 'FullBath', 'YearBuilt', 'Rooms']

X       = train_data[features]
y       = train_data.SalePrice
X_test  = test_data[features]

# Split for validation
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)
print(f"Train: {len(train_X)} | Val: {len(val_X)} | Test: {len(X_test)}")
```

## Step 2 — Train and Evaluate the Model

```python
model = RandomForestRegressor(n_estimators=100, random_state=1)
model.fit(train_X, train_y)

val_preds = model.predict(val_X)
mae = mean_absolute_error(val_y, val_preds)
print(f"Validation MAE: ${mae:,.0f}")
```

## Step 3 — Retrain on All Data and Predict the Test Set

```python
# Retrain on ALL available training data for the best predictions
final_model = RandomForestRegressor(n_estimators=100, random_state=1)
final_model.fit(X, y)

# Predict on the test set
test_preds = final_model.predict(X_test)
print(f"Generated {len(test_preds)} predictions.")
print(f"Predicted price range: ${test_preds.min():,.0f} – ${test_preds.max():,.0f}")
```

## Step 4 — Create the Submission File

```python
# Create submission DataFrame
output = pd.DataFrame({'Id': test_data.index, 'SalePrice': test_preds})
print("Submission preview:")
print(output.head())

# Save to CSV
output.to_csv('submission.csv', index=False)
print("\nSubmission file saved to: submission.csv")
print("Upload this file to Kaggle to see your competition score!")
```

## Tips for Improving Your Score

1. **Feature engineering** — Create new features from existing ones.
2. **Handle missing values** — Impute or drop missing data carefully.
3. **Try more powerful models** — Gradient Boosting (XGBoost, LightGBM).
4. **Hyperparameter tuning** — Use `GridSearchCV` or `RandomizedSearchCV`.
5. **Ensembling** — Combine predictions from multiple models.
